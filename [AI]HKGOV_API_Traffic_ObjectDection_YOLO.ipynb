# Train with coco dataset

```python
# Load a model
model = YOLO("yolo11n.pt")  # load a pretrained model (recommended for training)

# Train the model with coco dataset
results = model.train(data="coco8.yaml", epochs=100, imgsz=640)
```

# Image Detection

```python
from ultralytics import YOLO

# Define path to the image file
source = "/content/1.jpg"

# Run inference on the source
results = model(source, conf=0.3 )  # list of Results objects

# Ploting - Process results list
for result in results:
    boxes = result.boxes  # Boxes object for bounding box outputs
    masks = result.masks  # Masks object for segmentation masks outputs
    keypoints = result.keypoints  # Keypoints object for pose outputs
    probs = result.probs  # Probs object for classification outputs
    obb = result.obb  # Oriented boxes object for OBB outputs
    result.show()  # display to screen
    result.save(filename="result.jpg")  # save to disk
```

## Apply to stream image - Public camera

```python
import requests

url = "https://api.csdi.gov.hk/apim/dataquery/api/?id=td_rcd_1638952287148_39267&layer=traffic_camera_locations_en&limit=100&offset=0"

response = requests.get(url)

data = response.json()
data
```

```python
import pandas as pd

response = data

# Extract relevant data into a structured DataFrame
cameras = []
for feature in response['features']:
    lon, lat = feature['geometry']['coordinates']
    props = feature['properties']
    cameras.append({
        'OBJECTID': props['OBJECTID'],
        'district': props['district'],
        'description': props['description'],
        'region': props['region'],
        'key': props['key'],
        'url': props['url'],
        'lon': lon,
        'lat': lat
    })

df = pd.DataFrame(cameras)
df.head()
```

```python
# Find the target OBJECTID
filtered_df = df[df['description'].str.contains('harbour', case=False, na=False)]
filtered_df
```

```python
# Get the image_url of target camera
import requests

url = "https://api.csdi.gov.hk/apim/dataquery/api/?id=td_rcd_1638952287148_39267&layer=traffic_camera_locations_en&limit=100&offset=0"

response = requests.get(url)
if response.status_code == 200:
    data = response.json()
    for feature in data["features"]:
        if feature["properties"]["OBJECTID"] == 41:
            print(feature["properties"]["url"])
            image_url=feature["properties"]["url"]
            break
else:
    print(f"Error: {response.status_code}")
```

```python
from ultralytics import YOLO

# Define path to the image file
source = image_url

# Run inference on the source
results = model(source, conf=0.08 )  # list of Results objects

# Ploting - Process results list
for result in results:
    boxes = result.boxes  # Boxes object for bounding box outputs
    masks = result.masks  # Masks object for segmentation masks outputs
    keypoints = result.keypoints  # Keypoints object for pose outputs
    probs = result.probs  # Probs object for classification outputs
    obb = result.obb  # Oriented boxes object for OBB outputs
    result.show()  # display to screen
    result.save(filename="result.jpg")  # save to disk
```
